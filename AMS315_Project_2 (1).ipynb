```
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats

# 1. Set graph style
sns.set_theme(style="whitegrid")
plt.rcParams['figure.figsize'] = (10, 6)

# Load data
df = pd.read_csv('project2.csv')

# Check for missing values
print("--- Missing Values Check ---")
print(df.isnull().sum())

# Preview data structure
display(df.head())
display(df.info())
```

```
# 2. Visualize Target Variable (Price) Distribution
fig, ax = plt.subplots(1, 2, figsize=(15, 5))

# Original Price distribution
sns.histplot(df['price'], kde=True, ax=ax[0], color='blue')
ax[0].set_title('Distribution of House Prices')
ax[0].set_xlabel('Price')

# Log-transformed Price distribution
sns.histplot(np.log(df['price']), kde=True, ax=ax[1], color='green')
ax[1].set_title('Distribution of Log(Price)')
ax[1].set_xlabel('Log(Price)')

plt.tight_layout()
plt.show()

# Check skewness values
print("--- Skewness Check ---")
print(f"Price Skewness: {df['price'].skew():.2f}")
print(f"Log(Price) Skewness: {np.log(df['price']).skew():.2f}\n")
```

```
# 3. Pearson Correlation for Continuous Variables
# Select only continuous variables
continuous_vars = [
    'price', 'house_size', 'lot_size', 'size_of_the_yard',
    'number_of_rooms', 'number_of_bathrooms',
    'distance_to_the_beach', 'distance_to_malibu',
    'distance_to_santa_monica', 'distance_to_closest_freeway_entrance',
    'distance_to_shopping_center', 'school_district'
]

corr_matrix = df[continuous_vars].corr()

# Visualize correlation heatmap
plt.figure(figsize=(10, 8))
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm', fmt=".2f", vmin=-1, vmax=1)
plt.title('Correlation Matrix of Continuous Variables')
plt.show()

# Print correlation with price
print("--- Correlation with Price ---")
print(corr_matrix['price'].sort_values(ascending=False))
print("\n")
```

```
# 4. Premium Features Visualization & Welch's T-Test
# Categorical variables to analyze
categorical_vars = ['gated', 'solar_pannel', 'HOA']

fig, axes = plt.subplots(1, 3, figsize=(18, 5))

for i, var in enumerate(categorical_vars):
    # Visualize boxplots
    sns.boxplot(x=var, y='price', data=df, ax=axes[i], palette='Set2')
    axes[i].set_title(f'Price by {var}')

    # Perform Welch's T-test
    group1 = df[df[var] == 1]['price']
    group0 = df[df[var] == 0]['price']

    t_stat, p_val = stats.ttest_ind(group1, group0, equal_var=False)

    # Print results
    print(f"--- T-Test Results for '{var}' ---")
    print(f"T-statistic: {t_stat:.2f}, P-value: {p_val:.4e}")
    if p_val < 0.05:
        print(f"Result: There is a statistically significant difference in mean price based on '{var}'.\n")
    else:
        print(f"Result: No significant difference in mean price found based on '{var}'.\n")

plt.tight_layout()
plt.show()
```

```
# 5. One-Way ANOVA for variables with >2 categories
# Analysis of housing price differences based on the number of parking spots
var_anova = 'number_of_parking_spot'

# Boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(x=var_anova, y='price', data=df, palette='Pastel1')
plt.title(f'Price by {var_anova}')
plt.show()

# Separate data by group
groups = [group['price'].values for name, group in df.groupby(var_anova)]

# One-Way ANOVA
# H0: All the means are the same
# HA: At least one pair of mean is not the same
f_stat, p_val_anova = stats.f_oneway(*groups)

print(f"--- One-Way ANOVA Results for '{var_anova}' ---")
print(f"F-statistic: {f_stat:.2f}, P-value: {p_val_anova:.4e}")

if p_val_anova < 0.05:
    print(f"Result: There is a statistically significant difference in mean price based on {var_anova}.\n")
else:
    print(f"Result: There is no statistically significant difference in house price based on {var_anova}.\n")
```

```
import statsmodels.api as sm
from statsmodels.stats.diagnostic import het_breuschpagan
from statsmodels.stats.outliers_influence import variance_inflation_factor

# 6. Advanced Feature Selection & Final MLR Model Pipeline
print("Step A: Feature Selection & VIF Check")

# Step 1: Select optimal core features based on EDA results
selected_features = [
    'house_size',
    'number_of_bathrooms',
    'number_of_parking_spot',
    'distance_to_malibu',
    'distance_to_the_beach',
    'gated',
    'solar_pannel',
    'HOA'                      # Homeowners Association
]

X = df[selected_features]
X_const = sm.add_constant(X)
Y = df['price']

# Step 2: Check Multicollinearity using Variance Inflation Factor (VIF)
vif_data = pd.DataFrame()
vif_data["Feature"] = X_const.columns
vif_data["VIF"] = [variance_inflation_factor(X_const.values, i) for i in range(X_const.shape[1])]
print("--- Variance Inflation Factor (VIF) ---")
print(vif_data.to_string(index=False))
print("\n")


print("Step B: Original Multiple Linear Regression")

# Step 3: Fit the model with original 'Price' and check for heteroscedasticity
mlr_orig = sm.OLS(Y, X_const).fit()
print(mlr_orig.summary())

bp_test_orig = het_breuschpagan(mlr_orig.resid, mlr_orig.model.exog)
print(f"\nOriginal Model Breusch-Pagan P-value: {bp_test_orig[1]:.4e}")
if bp_test_orig[1] < 0.05:
    print("Result: Non-constant variance issue detected. Transformation needed.\n")


print("Step C: Log-Transformed Multiple Linear Regression")

# Step 4: Apply Natural Log transformation to 'Price' to resolve the issue
Y_log = np.log(df['price'])

# Fit the improved final model
mlr_log = sm.OLS(Y_log, X_const).fit(cov_type='HC3')
print(mlr_log.summary())

bp_test_log = het_breuschpagan(mlr_log.resid, mlr_log.model.exog)
print(f"\nLog-Transformed Model Breusch-Pagan P-value: {bp_test_log[1]:.4e}")
if bp_test_log[1] < 0.05:
    print("Result: BP test still shows p < 0.05 due to large sample size.")
    print("Action: Adopted Heteroscedasticity-Robust Standard Errors (HC3) to validate the model.\n")
else:
    print("Result: Variance is constant.\n")

# Step D: Residual Diagnostics Visualization
fig, axes = plt.subplots(2, 2, figsize=(16, 10))

# Residuals and QQ Plot for the original model
axes[0, 0].scatter(mlr_orig.fittedvalues, mlr_orig.resid, alpha=0.5)
axes[0, 0].axhline(0, color='red', linestyle='--')
axes[0, 0].set_title('Residuals vs Fitted (Original MLR)')
axes[0, 0].set_xlabel('Fitted Values')
axes[0, 0].set_ylabel('Residuals')

stats.probplot(mlr_orig.resid, dist="norm", plot=axes[0, 1])
axes[0, 1].set_title('QQ Plot (Original MLR)')

# Residuals and QQ Plot for the log-transformed model
axes[1, 0].scatter(mlr_log.fittedvalues, mlr_log.resid, alpha=0.5, color='green')
axes[1, 0].axhline(0, color='red', linestyle='--')
axes[1, 0].set_title('Residuals vs Fitted (Log-Transformed MLR)')
axes[1, 0].set_xlabel('Fitted Values (Log)')
axes[1, 0].set_ylabel('Residuals')

stats.probplot(mlr_log.resid, dist="norm", plot=axes[1, 1])
axes[1, 1].set_title('QQ Plot (Log-Transformed MLR)')

plt.tight_layout()
plt.show()
```
